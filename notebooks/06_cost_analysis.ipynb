# 06 · Cost variance

Standard prices are simulated analytical inputs.

```python
import pandas as pd
fact = pd.read_csv('../data/processed/fact_purchase_orders.csv')
print(fact.actual_spend.sum() - fact.expected_spend.sum())
print(fact.groupby('supplier_name')[['actual_spend','expected_spend']].sum().assign(var=lambda d: d.actual_spend-d.expected_spend).sort_values('var', ascending=False).head())
```
