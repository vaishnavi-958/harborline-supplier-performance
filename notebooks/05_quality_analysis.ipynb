```python
import pandas as pd
fact = pd.read_csv('../data/processed/fact_purchase_orders.csv')
print((fact.defective_quantity.sum() / fact.received_quantity.sum()))
print(fact.groupby('supplier_name').apply(lambda d: d.defective_quantity.sum() / max(d.received_quantity.sum(), 1)).sort_values(ascending=False).head())
```
